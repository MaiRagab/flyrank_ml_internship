*One metric you can defend. What number means 'good'?*

ROC-AUC is the primary evaluation metric because it measures how well the model distinguishes between customers who churn and those who do not.

A higher ROC-AUC score means better model performance. A value close to **1.0** is excellent, while **0.5** means the model performs no better than random guessing.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Machine learning can learn these patterns from the data and provide better recommendations than a simple if-statement based on one or two conditions.
